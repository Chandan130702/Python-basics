```python
'''
1. What is Simple Linear Regression?
--> Simple Linear Regression (SLR) is a statistical method that models the relationship between one independent variable (X) and one dependent variable (Y) using a straight line: Y = mX + c. It's used to understand how Y changes as X changes and to predict Y for new values of X.

2. What are the key assumptions of Simple Linear Regression?
--> Linearity — the relationship between X and Y is linear.
    Independence — observations (and their errors) are independent of one another.
    Homoscedasticity — the variance of residuals is constant across all values of X.
    Normality of residuals — errors are approximately normally distributed.

3. What does the coefficient m represent in the equation Y = mX + c?
--> The coefficient m represents the slope of the regression line. It dictates the expected change in the dependent variable (Y) for a one-unit increase in the independent variable (X).

4. What does the intercept c represent in the equation Y = mX + c?
--> The intercept c represents the starting point of the regression line on the y-axis. It is the predicted value of the dependent variable (Y) when the independent variable (X) is exactly zero.

5. How do we calculate the slope m in Simple Linear Regression?
--> The slope is mathematically calculated as the covariance between $X$ and $Y$ divided by the variance of $X$. The formula is:
    m = \frac{\sum (X_i - \bar{X})(Y_i - \bar{Y})}{\sum (X_i - \bar{X})^2}

6. What is the purpose of the least squares method in Simple Linear Regression?
--> The least squares method is an optimization technique used to find the best-fitting line. Its purpose is to minimize the sum of the squared differences (known as residuals or errors) between the actual observed values in the dataset and the values predicted by the linear model.

7. How is the coefficient of determination (R^2) interpreted in Simple Linear Regression?
--> R^2 is a metric ranging from 0 to 1 that represents the proportion of the variance in the dependent variable that is predictable from the independent variable. For example, an R^2 of 0.80 means that 80% of the variation in Y is explained by X.

8. What is Multiple Linear Regression?
--> It is an extension of simple linear regression that models the relationship between a continuous dependent variable and two or more independent variables.

9. What is the main difference between Simple and Multiple Linear Regression?
--> SLR uses exactly one predictor; MLR uses two or more. MLR can capture the combined effect of multiple factors on the outcome but introduces additional concerns like multicollinearity and the need to interpret each coefficient while holding others constant.

10. What are the key assumptions of Multiple Linear Regression?
--> It shares the same four assumptions as simple linear regression (Linearity, Independence, Homoscedasticity, Normality), with one crucial addition: No perfect multicollinearity (independent variables should not be highly correlated with each other).

11. What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model?
--> Heteroscedasticity occurs when the spread (variance) of the residuals is not constant across the range of predicted values. It invalidates the standard errors of the coefficients, making confidence intervals and p-values unreliable, which can lead to incorrect conclusions about feature significance.

12. How can you improve a Multiple Linear Regression model with high multicollinearity?
--> Remove or combine highly correlated predictors.
    Check Variance Inflation Factor (VIF) and drop variables with high VIF.
    Use dimensionality reduction (e.g., PCA).
    Apply regularization methods like Ridge or Lasso Regression.
    Increase sample size, or use domain knowledge to select more independent features.

13. What are some common techniques for transforming categorical variables for use in regression models?
--> One-hot encoding (creates a binary column per category).
    Dummy encoding (one-hot with n−1 categories to avoid the dummy variable trap).
    Label/ordinal encoding (for categories with a natural order).
    Target/mean encoding (replacing category with the mean of the target variable).
    Binary encoding (for high-cardinality categorical features).

14. What is the role of interaction terms in Multiple Linear Regression?
--> Interaction terms are used when the effect of one independent variable on the dependent variable changes depending on the value of another independent variable. They capture synergistic or antagonistic effects between features.

15. How can the interpretation of intercept differ between Simple and Multiple Linear Regression?
--> In simple regression, the intercept is the expected value of Y when the single X = 0. In multiple regression, the intercept is the expected value of Y when all independent variables in the model are simultaneously set to zero.

16. What is the significance of the slope in regression analysis, and how does it affect predictions?
--> In multiple regression, each slope (coefficient) represents the expected change in the target variable for a one-unit increase in that specific feature, assuming all other variables are held constant. It determines the weight and direction of that feature's influence on the final prediction.

17. How does the intercept in a regression model provide context for the relationship between variables?
--> It sets the baseline value of Y from which the effects of the predictors are added on. It anchors the regression line/plane and gives a reference point for interpreting how much the predictors shift Y away from that baseline.

18. What are the limitations of using R² as a sole measure of model performance?
--> It never decreases when you add more predictors, even irrelevant ones — encouraging overfitting.
    It doesn't indicate whether individual coefficients are statistically significant.
    It says nothing about prediction error in real units (unlike RMSE/MAE).
    It can be misleadingly high even when model assumptions are violated.
    It doesn't detect systematic bias or non-linearity missed by the model.
    (Adjusted R² partially addresses the first issue by penalizing extra predictors.)

19. How would you interpret a large standard error for a regression coefficient?
--> A large standard error indicates a high degree of uncertainty in the model's estimate of that coefficient. This usually results in a wide confidence interval and a high p-value, suggesting the variable may not be a statistically significant predictor of the target.

20. How can heteroscedasticity be identified in residual plots, and why is it important to address it?
--> It is typically identified by plotting the residuals against the predicted values; heteroscedasticity shows up as a "fan" or "cone" shape where the spread of residuals widens or narrows. It must be addressed to ensure the statistical tests determining feature importance are actually valid.

21. What does it mean if a Multiple Linear Regression model has a high R^2 but low adjusted R^2?
--> Adjusted R^2 penalizes the addition of variables that do not improve the model. If R^2 is high but Adjusted R^2 is low, it strongly indicates that the model contains too many useless predictors and is likely overfitting the training data.

22. Why is it important to scale variables in Multiple Linear Regression?
--> Predictors measured on very different scales (e.g., income in lakhs vs. age in years) can distort coefficient comparisons and make it hard to judge relative importance.
    Scaling (standardization/normalization) is essential for regularized regression (Ridge/Lasso) and gradient-based optimization, improving numerical stability and convergence.
    It ensures no single variable dominates the model purely because of its scale.

23. What is polynomial regression?
--> Polynomial regression is a form of regression analysis in which the relationship between the independent variable X and the dependent variable Y is modeled as an n-th degree polynomial in X.

24. How does polynomial regression differ from linear regression?
--> Linear regression fits a straight line and assumes a constant rate of change. Polynomial regression fits a curve by adding higher-power terms of X, allowing the model to capture non-linear patterns. Technically, polynomial regression is a special case of multiple linear regression, where the "multiple predictors" are just powers of the same original variable (X, X², X³, ...).

25. When is polynomial regression used?
--> When a scatter plot or residual plot from a linear model shows a clear curved pattern that a straight line fails to capture — i.e., the relationship between X and Y is non-linear.

26. What is the general equation for polynomial regression?
--> Y = b₀ + b₁X + b₂X² + b₃X³ + ... + bₙXⁿ + ε
    where n is the degree of the polynomial and ε is the error term.

27. Can polynomial regression be applied to multiple variables?
--> Yes. This is called multivariate polynomial regression, where powers and interaction terms of several variables are included, e.g.:
    Y = b₀ + b₁X₁ + b₂X₂ + b₃X₁² + b₄X₂² + b₅X₁X₂ + ...

28. What are the limitations of polynomial regression?
--> Overfitting at high degrees — the curve fits training data too closely, including noise.
    Sensitive to outliers, which can distort the curve significantly.
    Poor extrapolation — predictions beyond the range of training data can become extreme or unrealistic.
    Higher-degree terms introduce multicollinearity between polynomial features.
    Harder to interpret coefficients compared to simple linear terms.

29. What methods can be used to evaluate model fit when selecting the degree of a polynomial?
--> Cross-validation (e.g., k-fold) to see how well each degree generalizes.
    Examining residual plots for remaining patterns.
    Comparing train vs. test error (RMSE/MAE) to detect overfitting or underfitting.
    Applying the bias-variance tradeoff: low degree → high bias/underfitting; very high degree → high variance/overfitting.

30. Why is visualization important in polynomial regression?
--> Because polynomial models are prone to overfitting, visualizing the fitted curve over a scatter plot of the data allows you to quickly perform a "sanity check." You can instantly see if the curve naturally follows the trend or if it is erratically jumping around just to hit specific data points.

31. How is polynomial regression implemented in Python?
Typically with scikit-learn, by transforming X into polynomial features and then applying ordinary linear regression on those features:

import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Sample data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 6, 12, 20, 30])

# Transform X into polynomial features (degree = 2)
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# Fit linear regression on the transformed features
model = LinearRegression()
model.fit(X_poly, y)

# Predict
y_pred = model.predict(X_poly)

print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)

'''
```
